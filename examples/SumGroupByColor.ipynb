# Sum Group by Color
Save this notebook inside the Jupyter contents root. Restart the JupyterForExcel managed kernel after code changes. In Excel, reload the generated manifest and open Notebook Actions.
Example: enter 10, 20, 30, 40 in A1:A4. Fill A1 and A3 yellow, A2 white, and leave A4 with no fill. Select A1:A4 in the pane and Calculate. Expected totals: yellow 40; white 20; No fill 40.
Version one rejects conditional formatting, patterned fills and Excel error cells. It includes hidden/filtered cells and ignores text, blanks and booleans.

```python
from jupyterexcel import ribbon_function
from jupyterexcel.action_examples import group_sum_by_color
 

```

```python
   
@ribbon_function(
    name="SUM.GROUP.BY.COLOR",
    label="Sum Group by Color",
    button_text="Calculate",
    description="Group numeric cells by manual fill color. Text, blanks and booleans are ignored; hidden cells are included.",
    inputs={
        "data": {"source": "range", "type": "matrix", "label": "Data range",
                 "read": ["values", "format.fillColors"], "max_cells": 5000}
    },
    output={"type": "matrix", "destinations": "range"},
)
def sum_group_by_color(data):
    if 'debug_cache' not in globals():
        debug_cache = {}
    debug_cache ['color'] = data
    
    r= group_sum_by_color(data)
    debug_cache ['color_r'] = r
    return r
```

```python
import random
from datetime import date, datetime, timedelta


@ribbon_function(
    name="QUERY.PRICE.DATA",
    label="Query Price Data",
    button_text="Generate",
    description="Generate 100 rows of simulated weekday prices.",
    inputs={
        "ticker": {
            "source": "value",
            "type": "string",
            "label": "Ticker",
            "default": "MSFT",
        },
        "d": {
            "source": "value",
            "type": "string",
            "label": "Starting date (YYYY-MM-DD)",
            "default": "2026-09-01",
        },
    },
    output={"type": "matrix", "destinations": "range"},
)
def queryPriceData(ticker, d):
    if isinstance(d, str):
        d = date.fromisoformat(d)
    elif isinstance(d, datetime):
        d = d.date()

    r = []
    p = 118.0
    volatility = 0.2

    while len(r) < 100:
        if d.weekday() < 5:
            close = max(0.01, p + random.gauss(0, volatility))
            low = max(0.01, min(p, close) - random.uniform(0, volatility))
            high = max(p, close) + random.uniform(0, volatility)

            r.append([
                d.isoformat(),
                round(low, 2),
                round(high, 2),
                round(close, 2),
            ])
            p = close

        d += timedelta(days=1)

    return {"result": r}
```
