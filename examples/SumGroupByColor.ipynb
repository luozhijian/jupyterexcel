# Sum Group by Color
Save this notebook inside the Jupyter contents root. Restart the JupyterExcel managed kernel after code changes. In Excel, reload the generated manifest and open Notebook Actions.
Example: enter 10, 20, 30, 40 in A1:A4. Fill A1 and A3 yellow, A2 white, and leave A4 with no fill. Select A1:A4 in the pane and Calculate. Expected totals: yellow 40; white 20; No fill 40.
Version one rejects conditional formatting, patterned fills and Excel error cells. It includes hidden/filtered cells and ignores text, blanks and booleans.

```python
from jupyterexcel import ribbon_function
from jupyterexcel.action_examples import group_sum_by_color
 
debug_cache={}
@ribbon_function(
    name="SUM.GROUP.BY.COLOR",
    label="Sum Group by Color",
    button_text="Calculate",
    description="Group numeric cells by manual fill color. Text, blanks and booleans are ignored; hidden cells are included.",
    inputs={
        "data": {"source": "range", "type": "matrix", "label": "Data range",
                 "read": ["values", "format.fillColors"], "max_cells": 5000}
    },
    output={"type": "matrix", "destinations": "range"},
)
def sum_group_by_color(data):
    debug_cache ['color'] = data
    
    r= group_sum_by_color(data)
    debug_cache ['color_r'] = r
    return r
```
