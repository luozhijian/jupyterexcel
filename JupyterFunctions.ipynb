```python
@jupyter_function(name='Sum', description="Sum numbers and ranges",
    parameter_types={"values": "number"},
    parameter_dimensionality={"values": "matrix"},
    result_type="number",
)
def sum(*values):
    def sum_add(value):
        if isinstance(value, (list, tuple)):
            result = 0
            for item in value:
                result += sum_add(item)
            return result
        return value
    #to debug functions, cache input from Excel. Late please switch to related Kernel
    if 'debug_map' not in globals():
        globals()["debug_map"]={}
    debug_map ['Sum'] = values
    
    result = 0
    for value in values:
        result += sum_add(value)
    return result
```

```python
@jupyter_function(name="string_concatenate", description="concatenate string")
def string_concatenate(a, b):
    return  a  +  b 

    
@jupyter_function(name='string_join', description="Join multiple string together with delimiter, empty will be ignored",
    parameter_dimensionality={"values": "matrix"},
    result_type="string",
)
def string_join(delimiter, *values):
    result = []
    def _string_join(value):
        if not value :
            return 
        if isinstance(value, (list, tuple)):
            for item in value:
                _string_join(item)
        else :
            result.append( str(value) )
 
    for value in values:
        _string_join(value)
    return delimiter.join( result )  


@jupyter_function(name='string_join_trim', description="Join multiple string together with delimiter after remove leading and trailing space, empty will be ignored",
    parameter_dimensionality={"values": "matrix"},
    result_type="string",
)
def string_join_trim(delimiter, *values):
    result = []
    def _string_join(value):
        if not value :
            return 
        if isinstance(value, (list, tuple)):
            for item in value:
                _string_join(item)
        else :
            result.append( str(value).strip() )
 
    for value in values:
        _string_join(value)
    return delimiter.join( result )  
```

```python
@jupyter_function(name="Add", description="Add two values")
def add(a, b):
    return float(a) + float(b)
```

```python
@jupyter_function(
    name="Product",
    description="Multiply numbers and ranges",
    parameter_types={"values": "number"},
    parameter_dimensionality={"values": "matrix"},
    result_type="number",
)
def product(*values):
    def multiply(value):
        if isinstance(value, (list, tuple)):
            result = 1
            for item in value:
                result *= multiply(item)
            return result
        return value
    if 'debug_map' not in globals():
        globals()["debug_map"]={}
    debug_map ['product'] = values
    result = 1
    for value in values:
        result *= multiply(value)
    return result
```

```python
import numpy as np

#predict a ball height when throwing
@jupyter_function(
        name="PredictHeight",
    description="Predict Ball Height when it is in air",
    parameter_types={"times": "number", "heights":"number", "query_times":"number"},
    parameter_dimensionality={"times": "matrix", "heights":"matrix", "query_times":"matrix"},
    result_type="number",
    result_dimensionality="matrix",
)
def predict_height(times, heights, query_times):
    """Fit a quadratic and predict height at the requested times."""
    t = np.asarray(times, dtype=float).ravel()
    h = np.asarray(heights, dtype=float).ravel()
    q = np.asarray(query_times, dtype=float).ravel()

    coefficients = np.polyfit(t, h, 2)
    return np.polyval(coefficients, q).reshape(-1, 1).tolist()
```
