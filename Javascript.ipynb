```typescript
const sales = [
  { product: "Keyboard", quantity: 4, price: 75 },
  { product: "Mouse", quantity: 10, price: 25 },
  { product: "Monitor", quantity: 3, price: 240 },
];

const revenue = sales.map(item => ({
  ...item,
  revenue: item.quantity * item.price,
}));

console.table(revenue);
```

```typescript
const total = revenue.reduce((sum, item) => sum + item.revenue, 0);
const best = revenue.reduce((a, b) => a.revenue > b.revenue ? a : b);

Deno.jupyter.md`
## Sales Summary
- **Total revenue:** $${total.toLocaleString("en-US")}
- **Top product by revenue:** ${best.product}
`;
```
