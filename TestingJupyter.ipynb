```python
import datetime
from jupyterexcel.utils import  ribbon_function,get_ribbon_functions,jupyter_function

```

```python
@jupyter_function(name='sum', description='sum multiple numbers')
def sum(a, b, c):
    return float(a) + float(b) +float(c)


@jupyter_function(name="ADD", description="Add two values")
def add(a, b):
    return float(a) + float(b)

@jupyter_function(name="string_concatenate", description="concatenate string")
def string_concatenate(a, b):
    return  a  +  b 
```

```python
@jupyter_function(
    name="PRODUCT",
    description="Multiply numbers and ranges",
    parameter_types={"values": "number"},
    parameter_dimensionality={"values": "matrix"},
    result_type="number",
)
def product(*values):
    def multiply(value):
        if isinstance(value, (list, tuple)):
            result = 1
            for item in value:
                result *= multiply(item)
            return result
        return value

    result = 1
    for value in values:
        result *= multiply(value)
    return result
```

```python
@jupyter_function(name='returnArray', description='return array sample' ,
        result_type="number",
        result_dimensionality="matrix",
    )
def returnArray(a, b, c):
    return [[float(a) *2 , float(b)*2,float(c)*2] ,
            [float(a) *3 , float(b)*3,float(c)*3] ,
            [float(a) *4 , float(b)*4,float(c)*4] ]
```

```python
def returnString(a, b=0, c=0):
    return "[%lf , %lf,'' %lf]"%(float(a) , float(b),float(c))
```

```python
def return2DMixArray(a, b=0, c=0):
    return [[
         float(b),float(c), True], [ float(b)*2,float(c)*2, datetime.datetime.now()] ]
```

```python
@ribbon_function(name="return2DArray",  a="D6", b="Sheet2!C4", return_value="D24:F25")
def return2DArray(a, b=0, c=0):
    return [[float(a) , float(b),float(c)], [float(a)*2, float(b)*2,float(c)*2]]
```
